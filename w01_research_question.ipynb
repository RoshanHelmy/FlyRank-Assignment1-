## 2. The question: decision, action, cost of a wrong call

Which content pages should be prioritized for review and refresh?

### Action
Content editors and SEO teams review the highest-ranked pages first and decide whether to update, expand, or improve the content.

### Cost of a Wrong Recommendation
A false positive may waste editorial time on a page that does not require updating. A false negative may cause a truly declining page to continue losing visibility and traffic before it is reviewed.

Machine learning can combine multiple observable signals, such as impressions, CTR, content age, and page position, to identify patterns that are difficult to capture using simple hand-written rules alone.

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains 30,000 anonymized content pages, providing enough examples to study patterns related to declining content.

Approximately 54.2% of the pages are labeled as declining. Additionally, the correlation between search volume and impressions is approximately 0.001, suggesting that search volume alone is not a strong predictor of page traffic in this dataset.

```python
import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")

print("Number of pages:", len(df))

declining_rate = (df["trend_direction"] == "down").mean()
print(f"Declining rate: {declining_rate:.3f}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
```
